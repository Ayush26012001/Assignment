
# Regression Concepts – Questions & Answers

---



**Q1. What is Simple Linear Regression?**  
Simple Linear Regression is a statistical technique used to model the relationship between one independent variable (X) and one dependent variable (Y) by fitting a straight line.



**Q2. What are the key assumptions of Simple Linear Regression?**  
- Linearity  
- Independence of errors  
- Homoscedasticity  
- Normality of errors  
- No significant outliers



**Q3. What does the coefficient m represent in Y = mX + c?**  
m represents the slope, i.e., the change in Y for a one-unit change in X.



**Q4. What does the intercept c represent in Y = mX + c?**  
c is the value of Y when X = 0.



**Q5. How do we calculate the slope m in Simple Linear Regression?**  
m = Cov(X, Y) / Var(X)



**Q6. What is the purpose of the least squares method?**  
To minimize the sum of squared residuals between observed and predicted values.



**Q7. How is R² interpreted?**  
R² represents the proportion of variance in Y explained by X.



**Q8. What is Multiple Linear Regression?**  
It models the relationship between one dependent variable and multiple independent variables.



**Q9. Main difference between Simple and Multiple Linear Regression?**  
Simple uses one predictor; Multiple uses more than one.



**Q10. Key assumptions of Multiple Linear Regression?**  
- Linearity  
- Independence  
- Homoscedasticity  
- Normality  
- No multicollinearity



**Q11. What is heteroscedasticity?**  
It occurs when the variance of errors is not constant, leading to inefficient estimates.



**Q12. How to improve a model with high multicollinearity?**  
- Remove correlated variables  
- Use PCA  
- Apply Ridge/Lasso regression



**Q13. Techniques to transform categorical variables?**  
- One-hot encoding  
- Label encoding  
- Target encoding



**Q14. Role of interaction terms?**  
They capture combined effects of variables on the dependent variable.



**Q15. Intercept interpretation difference?**  
In Multiple Regression, intercept is the expected Y when all Xs are zero.



**Q16. Significance of slope?**  
It shows direction and strength of relationship and affects predictions.



**Q17. Limitations of R²?**  
- Doesn’t indicate causation  
- Increases with more predictors  
- Ignores overfitting



**Q18. Interpretation of large standard error?**  
Indicates unreliable coefficient estimates.



**Q19. What is Polynomial Regression?**  
A regression model where predictors are raised to polynomial powers.



**Q20. When is Polynomial Regression used?**  
When data shows a nonlinear relationship.



**Q21. How does intercept provide context?**  
It defines baseline Y value for comparison.



**Q22. Identifying heteroscedasticity in residual plots?**  
Funnel or pattern-shaped residuals indicate heteroscedasticity.



**Q23. High R² but low Adjusted R² means?**  
Model may include unnecessary predictors.



**Q24. Why scale variables?**  
To ensure equal contribution and numerical stability.



**Q25. Polynomial vs Linear Regression?**  
Polynomial fits curves; linear fits straight lines.



**Q26. General equation for Polynomial Regression?**  
Y = a₀ + a₁X + a₂X² + ... + aₙXⁿ



**Q27. Can Polynomial Regression be applied to multiple variables?**  
Yes, by including polynomial terms for each variable.



**Q28. Limitations of Polynomial Regression?**  
- Overfitting  
- Poor extrapolation  
- Complex interpretation



**Q29. Methods to evaluate polynomial degree?**  
- Cross-validation  
- Adjusted R²  
- AIC/BIC



**Q30. Why is visualization important?**  
It helps detect nonlinearity and overfitting.
